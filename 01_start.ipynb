# Start project

The goal of the project is to create a dashboard that tracks the user's workout progress and provides insights into their training habits. The dashboard will be built using Python and will utilize various libraries for data analysis and visualization.

[source](https://www.geeksforgeeks.org/data-analysis/six-steps-of-data-analysis-process/)


## Focus

>The goal of the notebooks is to create an overview on the structure of the data and if there are any correlation between columns.

## Questions

### Consistency

1. How often does the user train per week?
2. On which days does the user train the most?
3. How does the workout frequency change over time?
4. At what time does the user workout the most?

### Volume and Strength

5. Does the user gets stronger over time?
6. Which exercise shows the most growth overtime?
7. Are there any plateaus in the progress (strength)?
8. Does the volume for the exercises increase overtime?
9. Are there any spikes or drops in strength/volume?

### Correlation Questions

10.  Does the time of the workout affects the strength of the user?
11.  Does increased volume of an exercise improve the overal strenth overtime?
12.  Does more workouts each week correlate with more strength gain?
13.  Does the user perform better on certain days of the week?
14.  Does the user perform better at certain times of the day?
