## Dataset overview

DataFrame `data` is loaded from `cleaned_data.csv`. At export time the frame had **4,158 rows** and **11 columns**, with no missing values in any column.

## Column reference

| Column | Dtype | Non-null | Description |
|--------|-------|----------|-------------|
| `title` | object | 4158 | Workout session title from the app export. |
| `date` | object | 4158 | Workout or log timestamp (string). |
| `duration` | object | 4158 | Workout duration as stored in the export. |
| `exercise` | object | 4158 | Exercise name; rows are rep-level logs from Lyfta. |
| `weight` | float64 | 4158 | Load used for the rep (same units as in the app). |
| `reps` | float64 | 4158 | Repetition index or count within the set. |
| `set_type` | object | 4158 | Set type from the app (e.g. warm-up vs working set). |
| `date_short` | object | 4158 | Calendar date (`YYYY-MM-DD`) for joining with body weight. |
| `body_weight` | float64 | 4158 | User body weight on that date (from joined scale data). |
| `primary_muscle` | object | 4158 | Primary muscle group mapped from the exercise name. |
| `secondary_muscle` | object | 4158 | Secondary muscle group mapped from the exercise name. |
